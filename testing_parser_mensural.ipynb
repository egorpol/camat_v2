```python
# Imports
from IPython.display import display

from camat.parser_registry import parse_files

# Optional: auto summary via parser_registry.parse_files(..., print_parsed_summary=...)
# False disables it; a dict forwards kwargs to parser_utils.print_parsed_summary.
PRINT_PARSED_SUMMARY = True

# ---------------------------------------------------------------------
# Parameter section used in this mensural notebook
# ---------------------------------------------------------------------
FILTER_ZERO_DURATION = True          # filter out notes with duration 0 (grace notes)
ADJUST_FRACTIONAL_DURATION = True    # rounds Duration, Local Onset, and Global Onset to 3 decimals
PARSE_ENHARMONIC = True              # add 'Pitch Enharmonic' column grounded in source notation
DEDUPE_WEAKER_TEXT_EVENTS = False    # drops weaker duplicate text-like events when an anchored copy also exists

PLOTTING_BACKEND = 'bokeh'           # 'plt' | 'bokeh' | 'none'
SHOW_HOVER = True                    # enable hover tool (Bokeh only)
HOVER_FIELDS = [
    'measure', 'local_onset', 'global_onset', 'duration',
    'pitch', 'pitch_enharmonic', 'midi', 'voice', 'xml_id'
]

DISPLAY_PREVIEW_DF_PITCH = True      # show preview for parsed pitch dataframe
DISPLAY_PREVIEW_DF_EVENTS = True     # show preview for parsed event dataframe
PREVIEW_ROWS = 10                    # number of rows to show in preview
CLEANUP_REMOTE = True                # if source is a URL, delete temp file after parsing
RETURN_PLOTS = False                 # include plot objects in results under 'plot'
PLOT_SIZE_X = 900                    # plot width in pixels
PLOT_SIZE_Y = 600                    # plot height in pixels
ZOOM_DRAG_DIM = 'both'               # 'both' | 'width' | 'height' (Bokeh only)
ZOOM_WHEEL_DIM = 'width'             # 'both' | 'width' | 'height' (Bokeh only)
SHOW_PROGRESS = True                 # tqdm progress bar for multiple files
PROGRESS_DESC = None                 # custom progress label, or None

# Mensural parsing / display options
ALIGN_ACCIDENT_SCHEMA = True         # normalize accidental spellings in parsed pitch columns
COLORIZE_VOICES = True               # color-code notes by voice/part in piano roll
PALETTE = 'Category20'               # e.g. 'Category10', 'Category20', 'Set3', or list of colors
INCLUDE_XML_IDS = True               # for MEI, include xml:id as 'xml_id'
PLOT_PARSED_BARLINES_WITH_VOICE_COLORING = True  # overlay parsed MEI barlines using voice-matched colors

# List of file sources (URLs or local paths)
FILE_SOURCES = [
    'https://github.com/annplaksin/mei-sample-encodings/raw/refs/heads/main/Dufay-M-SeLaFace/Capp.Sist.%2014,%2027v-28r_MEI_Mensural_corr.mei',
    #'https://raw.githubusercontent.com/annplaksin/mei-sample-encodings/refs/heads/main/Dufay-SalveRegina/SalveReginaMisericordiae_Dufay_MEI_Mensural.mei',
    #'https://github.com/annplaksin/mei-sample-encodings/raw/refs/heads/main/Josquin_M-AveMarisStella_KyrieI/Jena3/M-AveMArisStella_KyrieI_Josquin_Jena3.mei',
]

# Parse and visualize
results, dfs_by_name, df_processed = parse_files(
    FILE_SOURCES,
    parsing_backend='mensural',
    dedupe_weaker_text_events=DEDUPE_WEAKER_TEXT_EVENTS,
    print_parsed_summary=PRINT_PARSED_SUMMARY,
    filter_zero_duration=FILTER_ZERO_DURATION,
    adjust_fractional_duration=ADJUST_FRACTIONAL_DURATION,
    parse_enharmonic=PARSE_ENHARMONIC,
    backend=PLOTTING_BACKEND,
    show_hover=SHOW_HOVER,
    hover_fields=HOVER_FIELDS,
    display_preview_df_pitch=DISPLAY_PREVIEW_DF_PITCH,
    display_preview_df_events=DISPLAY_PREVIEW_DF_EVENTS,
    preview_rows=PREVIEW_ROWS,
    cleanup_remote=CLEANUP_REMOTE,
    return_plots=RETURN_PLOTS,
    plot_width=PLOT_SIZE_X,
    plot_height=PLOT_SIZE_Y,
    zoom_drag_dim=ZOOM_DRAG_DIM,
    zoom_wheel_dim=ZOOM_WHEEL_DIM,
    show_progress=SHOW_PROGRESS,
    progress_desc=PROGRESS_DESC,
    align_accident_schema=ALIGN_ACCIDENT_SCHEMA,
    colorize_voices=COLORIZE_VOICES,
    palette=PALETTE,
    include_xml_ids=INCLUDE_XML_IDS,
    plot_parsed_barlines_with_voice_coloring=PLOT_PARSED_BARLINES_WITH_VOICE_COLORING,
)
```

```python
PITCH_KEYS = [name for name in dfs_by_name if name.endswith("_pitch")]
EVENT_KEYS = [name for name in dfs_by_name if name.endswith("_events")]
ACTIVE_PITCH_KEY = PITCH_KEYS[0] if PITCH_KEYS else None
ACTIVE_EVENTS_KEY = EVENT_KEYS[0] if EVENT_KEYS else None
print("Pitch key:", ACTIVE_PITCH_KEY)
print("Event key:", ACTIVE_EVENTS_KEY)
list(dfs_by_name)
```

```python
from camat import display_filtered_piano_roll

# --- Configuration ---
SOURCE_DF = dfs_by_name[ACTIVE_PITCH_KEY]             # which dataframe to use (df1, df2, etc.)
MEASURE_RANGE = None        # (start, end) inclusive, or None
ONSET_RANGE = None       # (start, end) exclusive at end [start, end), or None
VOICE_QUERY = ["Contra - Voice 1"]  # regex string OR list of exact voice names, or None
PITCH_RANGE = None          # pitch range filter; accepts MIDI or names, e.g. (60, 72) or "g3-g5"

# Display options for the selection DataFrame
DISPLAY_SELECTION = True    # True = show selection, False = hide
DISPLAY_MODE = "head"       # 'head' (default), 'all', or 'none'
DISPLAY_MAX_ROWS = 20       # how many rows to show in 'head' mode

# Plotting options
PLOTTING_BACKEND = (PLOTTING_BACKEND.lower() if 'PLOTTING_BACKEND' in globals() else 'plt')
PLOT_WIDTH = (PLOT_SIZE_X if 'PLOT_SIZE_X' in globals() else None)
PLOT_HEIGHT = (PLOT_SIZE_Y if 'PLOT_SIZE_Y' in globals() else None)

# Optional: pass through `results` if you have it from earlier parsing steps
_results_arg = results if 'results' in globals() else None

EVENTS_DF = dfs_by_name[ACTIVE_EVENTS_KEY]

selection = display_filtered_piano_roll(
    SOURCE_DF,
    measure_range=MEASURE_RANGE,
    onset_range=ONSET_RANGE,
    voice_query=VOICE_QUERY,
    pitch_range=PITCH_RANGE,
    results=_results_arg,
    events_df=EVENTS_DF,
    plotting_backend=PLOTTING_BACKEND,
    plot_width=PLOT_WIDTH,
    plot_height=PLOT_HEIGHT,
    zoom_drag_dim=ZOOM_DRAG_DIM if 'ZOOM_DRAG_DIM' in globals() else None,
    zoom_wheel_dim=ZOOM_WHEEL_DIM if 'ZOOM_WHEEL_DIM' in globals() else None,
    show_measure_lines=False,
    plot_parsed_barlines_with_voice_coloring=True,
    display_selection=DISPLAY_SELECTION,
    display_mode=DISPLAY_MODE,
    display_max_rows=DISPLAY_MAX_ROWS,
    colorize_voices=COLORIZE_VOICES if 'COLORIZE_VOICES' in globals() else True,
    palette=PALETTE if 'PALETTE' in globals() else 'Category20',
    preserve_voice_color_mapping=True,
)
```

```python
url = FILE_SOURCES[0]

from camat import (
    vrv_set_options, vrv_load_from_url, vrv_load_from_file,
    vrv_render_page, vrv_display_svg, vrv_debug_info
)
vrv_set_options(pageWidth=12000, pageHeight=1200, scale=40, breaks="none", removeIds=False, xmlIdSeed=0)

# Loading does not auto-render; explicitly render/display after load.
if isinstance(url, str) and url.startswith(("http://", "https://")):
    pages = vrv_load_from_url(url)
else:
    pages = vrv_load_from_file(url)

print(f"Verovio loaded pages: {pages}")
if pages > 0:
    svg = vrv_render_page(1)
    vrv_display_svg(svg)
else:
    print("No pages rendered. Debug info:", vrv_debug_info())
```
