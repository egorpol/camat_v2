```python
# Imports
from IPython.display import display

from camat.parser_registry import parse_files

# ---------------------------------------------------------------------
# Scope: this notebook is for COMMON-NOTATION scores only.
# Mensural notation lives in `testing_parser_mensural.ipynb` (or call
# `parse_files(..., parsing_backend='mensural')` directly). All mensural
# preprocessing paths are explicitly disabled below so they never run.
# ---------------------------------------------------------------------

# Backend selection ('music21' | 'partitura')
PARSING_BACKEND = 'partitura'

# Optional: auto summary via parser_registry.parse_files(..., print_parsed_summary=...)
# False disables it; a dict forwards kwargs to parser_utils.print_parsed_summary.
PRINT_PARSED_SUMMARY = True

# ---------------------------------------------------------------------
# Parameter section used in this notebook
# ---------------------------------------------------------------------
FILTER_ZERO_DURATION = True          # filter out notes with duration 0 (grace notes)
ADJUST_FRACTIONAL_DURATION = True    # rounds Duration, Local Onset, and Global Onset to 3 decimals
PARSE_ENHARMONIC = True              # add 'Pitch Enharmonic' column grounded in source notation
DEDUPE_WEAKER_TEXT_EVENTS = False    # drops weaker duplicate text-like events when an anchored copy also exists

PLOTTING_BACKEND = 'bokeh'           # 'plt' | 'bokeh' | 'none'
SHOW_MEASURE_LINES = False            # draw vertical measure lines at measure starts
MEASURE_LINE_COLOR = 'black'         # color for measure lines
SHOW_HOVER = True                    # enable hover tool (Bokeh only)
HOVER_FIELDS = [
    'measure', 'local_onset', 'global_onset', 'duration',
    'pitch', 'pitch_enharmonic', 'midi', 'voice', 'xml_id'
]

DISPLAY_PREVIEW_DF_PITCH = True      # show preview for parsed pitch dataframe
DISPLAY_PREVIEW_DF_EVENTS = True     # show preview for parsed event dataframe
PREVIEW_ROWS = 20                    # number of rows to show in preview
CLEANUP_REMOTE = True                # if source is a URL, delete temp file after parsing
RETURN_PLOTS = False                 # include plot objects in results under 'plot'
PLOT_SIZE_X = 900                    # plot width in pixels
PLOT_SIZE_Y = 600                    # plot height in pixels
ZOOM_DRAG_DIM = 'both'               # 'both' | 'width' | 'height' (Bokeh only)
ZOOM_WHEEL_DIM = 'width'             # 'both' | 'width' | 'height' (Bokeh only)
SHOW_PROGRESS = True                 # tqdm progress bar for multiple files
PROGRESS_DESC = None                 # custom progress label, or None

QUIET_NATIVE_WARNINGS = False        # suppress native backend warnings

# Parallelism + remote source caching
PARALLEL_N_JOBS = -1                 # 1 = serial; >1 uses a thread pool; -1 = os.cpu_count()
USE_REMOTE_CACHE = True              # cache URL downloads under ~/.cache/camat/downloads
REMOTE_CACHE_DIR = None              # override cache dir (None uses default or $CAMAT_DOWNLOAD_CACHE_DIR)

# Cross-backend options
STRIP_TIES = True                    # reject ties in music21 backend only
ALIGN_ACCIDENT_SCHEMA = True         # harmonize accidental schema across backends
COLORIZE_VOICES = True               # color-code notes by voice/part in piano roll
PALETTE = 'Category20'               # e.g. 'Category10', 'Category20', 'Set3', or list of colors
INCLUDE_XML_IDS = True               # for MEI, include xml:id as 'xml_id'

# MEI / parser fallback handling
TRY_VEROVIO_MEI_CONVERSION = True        # retry unsupported MEI structures via Verovio conversion
PLOT_PARSED_BARLINES_WITH_VOICE_COLORING = True  # overlay parsed MEI barlines using voice-matched colors
ALLOW_MUSIC21_FALLBACK = False           # strict partitura flow when False; no music21 fallback

# List of file sources (URLs or local paths)
# NOTE: only common-notation MusicXML / MEI / Kern sources belong here.
# For mensural MEI use `testing_parser_mensural.ipynb`.
FILE_SOURCES = [
    # 'https://analyse.hfm-weimar.de/database/02/PrJode_Jos0302_COM_1-5_MissaDapac_002_00006.xml',
    # 'https://raw.githubusercontent.com/humdrum-tools/bach-wtc-fugues/refs/heads/master/kern/wtc1f04.krn',
    # 'https://raw.githubusercontent.com/humdrum-tools/bach-wtc/refs/heads/main/kern/wtc1f22.krn',
    #  'https://raw.githubusercontent.com/piasteuck/winterreise-analysis/refs/heads/main/Schubert_Winterreise_Dataset_v2-0/01_RawData/score_musicxml/Schubert_D911-20.xml',
    'https://raw.githubusercontent.com/music-encoding/sample-encodings/main/MEI_5.0/Music/Complete_examples/Bach-JS_Ein_feste_Burg.mei',
    'https://raw.githubusercontent.com/music-encoding/sample-encodings/main/MEI_3.0/Music/Complete_examples/Mozart_Fuge_G_minor.mei',
    'https://raw.githubusercontent.com/trompamusic-encodings/Beethoven_Op31_No3_HenleUrtext/refs/heads/master/Beethoven_Op31_No3_3-HenleUrtext.mei',
]

# Parse and visualize
results, dfs_by_name, df_processed = parse_files(
    FILE_SOURCES,
    parsing_backend=PARSING_BACKEND,
    quiet_native_warnings=QUIET_NATIVE_WARNINGS,
    dedupe_weaker_text_events=DEDUPE_WEAKER_TEXT_EVENTS,
    print_parsed_summary=PRINT_PARSED_SUMMARY,
    filter_zero_duration=FILTER_ZERO_DURATION,
    adjust_fractional_duration=ADJUST_FRACTIONAL_DURATION,
    parse_enharmonic=PARSE_ENHARMONIC,
    backend=PLOTTING_BACKEND,
    show_measure_lines=SHOW_MEASURE_LINES,
    measure_line_color=MEASURE_LINE_COLOR,
    show_hover=SHOW_HOVER,
    hover_fields=HOVER_FIELDS,
    display_preview_df_pitch=DISPLAY_PREVIEW_DF_PITCH,
    display_preview_df_events=DISPLAY_PREVIEW_DF_EVENTS,
    preview_rows=PREVIEW_ROWS,
    cleanup_remote=CLEANUP_REMOTE,
    return_plots=RETURN_PLOTS,
    plot_width=PLOT_SIZE_X,
    plot_height=PLOT_SIZE_Y,
    zoom_drag_dim=ZOOM_DRAG_DIM,
    zoom_wheel_dim=ZOOM_WHEEL_DIM,
    show_progress=SHOW_PROGRESS,
    progress_desc=PROGRESS_DESC,
    strip_ties=STRIP_TIES,
    align_accident_schema=ALIGN_ACCIDENT_SCHEMA,
    colorize_voices=COLORIZE_VOICES,
    palette=PALETTE,
    include_xml_ids=INCLUDE_XML_IDS,
    try_verovio_mei_conversion=TRY_VEROVIO_MEI_CONVERSION,
    plot_parsed_barlines_with_voice_coloring=PLOT_PARSED_BARLINES_WITH_VOICE_COLORING,
    allow_music21_fallback=ALLOW_MUSIC21_FALLBACK,
    use_remote_cache=USE_REMOTE_CACHE,
    remote_cache_dir=REMOTE_CACHE_DIR,
    n_jobs=PARALLEL_N_JOBS,
    # --- common-notation only: explicitly disable every mensural code path ---
    # Keeps the partitura backend on its common-notation fast path regardless of
    # its defaults, and silences the "Skipping mensural duration/meter
    # preprocessing" log. For mensural scores, use testing_parser_mensural.ipynb.
    normalize_mensural_durations=False,
    inject_missing_meter_signature=False,
    prefer_verovio_for_mensural=False,
    use_verovio_mensural_timing=False,
)
```

### Annotation behavior

- `tstamp` annotations create `<annot>` elements anchored to a measure/time range (`measure_index`, `staff`, `tstamp`, `tstamp2`). Re-running the same spec updates the existing annotation when it has an `xml_id`; `vrv_process_annotations(...)` now also assigns stable internal ids when `xml_id` is omitted, but using an explicit `xml_id` is still the clearest option.
- `plist` annotations point to existing score elements by MEI `xml:id` (for example `#note-id`). The `plist` controls which rendered notes are highlighted, while the annotation's own `xml_id` controls whether the `<annot>` entry is updated or duplicated on later runs.

```python
# Refactored pitch distribution using py_scripts.analysis_utils
from camat import display_pitch_distribution

# Configuration for this cell (all lowercase options)
#
# PITCH_X_AXIS controls which pitch representation is used on the X axis:
#   - 'midi': use numeric MIDI values. If 'MIDI' column is missing, it is derived from names.
#   - 'pitch real' (synonyms: 'pitch', 'real'): use real pitch names corresponding to actual semitone values;
#       if only MIDI is present, names are derived from MIDI. Octave is included in the label.
#   - 'pitch enharmonic' (synonyms: 'pitch by name', 'pitch name', 'name', 'pitch_enharmonic', 'pitch-enharmonic'):
#       use written/enharmonic pitch names as notated (prefers 'Pitch Enharmonic' column),
#       preserving the spelled accidentals and octave, even if enharmonically different from real pitch.
PITCH_X_AXIS = 'pitch enharmonic'
#
# ORDER_X_AXIS_BY controls sort order of the X categories:
#   - 'midi' (synonyms: 'pitch real', 'pitch', 'real'):
#       sort by real MIDI number ascending (identical for all synonyms).
#   - 'pitch by octave real' (synonyms: 'pitch by octave', 'pitch octave', 'octave pitch', 'octave real'):
#       sort by REAL octave (computed from MIDI), then by pitch class within each octave.
#       This cares about actual pitch height. Example: B##3 resolves to its real octave (4); Cbb4 resolves to 3.
#   - 'pitch by octave enharmonic' (synonyms: 'octave enharmonic', 'octave name'):
#       sort by WRITTEN octave parsed from the name (ignoring semitone shifts of accidentals),
#       then by letter and accidental. Example: B##3 stays in octave 3; Cbb4 stays in octave 4.
#   - 'pitch by name' (synonyms: 'pitch enharmonic', 'enharmonic', 'enharmonic pitch', 'pitch_enharmonic'):
#       sort by letter (C..B) and accidental rank (bbb < bb < b < natural < # < ## < ###),
#       ignoring octave; ties are stabilized by the full label.
ORDER_X_AXIS_BY = 'midi'

# Normalize the individual pitch distribution to the range 0-1
# Each source DataFrame is normalized independently before the combined plot is built, 
# so in multi-source mode every series sums to 1.0. That means a source with 30 notes and a source with 
# 300 notes are both plotted as per-source proportions and can be compared directly.
NORMALIZE = True
FLOAT_FORMAT = '3f'   # same as '.3f'

# Plotting backend
PLOTTING_BACKEND = 'bokeh'        # 'plt' | 'bokeh' | 'none'

# Plot sizing for this cell (universal for both backends)
PLOT_SIZE_X = 1200
PLOT_SIZE_Y = 450

# Bokeh hover config
BOKEH_HOVER_VALUES = True

# Bar color for this plot
PITCH_BAR_COLOR = '#4682B4' # bar color(s) for pitch distribution plot(s)

# Choose the DataFrame to visualize
source = (selection, dfs_by_name["00_bach_js_ein_feste_burg_pitch"], dfs_by_name["01_mozart_fuge_g_minor_pitch"])  # e.g., df1, results[0]['df'], dfs_by_name['00_some_name'] or a selection from df

counts_df = display_pitch_distribution(
    source,
    source_labels=['bach short', 'bach full', 'mozart full'],
    pitch_axis=PITCH_X_AXIS,
    order_axis_by=ORDER_X_AXIS_BY,
    backend=(PLOTTING_BACKEND.lower() if 'PLOTTING_BACKEND' in globals() else 'plt'),
    plot_width=PLOT_SIZE_X,
    plot_height=PLOT_SIZE_Y,
    show_hover=('BOKEH_HOVER_VALUES' in globals() and BOKEH_HOVER_VALUES),
    show_table=True,
    bar_color=PITCH_BAR_COLOR,
    normalize=NORMALIZE,
    float_format=FLOAT_FORMAT,
)
```
