# Verovio Conversion Showcase

This notebook demonstrates how to use the `vrv_convert_to_mei` function to convert MusicXML and Humdrum files to MEI using the Verovio engine.

We will also verify the conversion by rendering the resulting MEI.


```python
import os
import sys

# Add the current directory to path so we can import py_scripts
if "." not in sys.path:
    sys.path.append(".")

from py_scripts.verovio_render import (
    vrv_convert_to_mei, 
    vrv_set_options, 
    vrv_render_page, 
    vrv_display_svg
)

# Set some basic render options
vrv_set_options(pageWidth=1000, pageHeight=1000, scale=40)
```

## 1. Convert Humdrum (Kern) to MEI

We will use a Beethoven Sonata movement from the Craig Sapp repository.


```python
humdrum_url = "https://raw.githubusercontent.com/craigsapp/beethoven-piano-sonatas/master/kern/sonata29-1.krn"

print("Converting Humdrum file...")
# Convert directly from URL
mei_output_humdrum = vrv_convert_to_mei(humdrum_url, is_url=True)

print(f"Conversion successful. Length of MEI: {len(mei_output_humdrum)} characters")
print("First 500 characters of MEI:")
print(mei_output_humdrum[:500])
```

### Verify Render (Humdrum Source)
Since `vrv_convert_to_mei` loads the data into the toolkit, we can render the first page immediately.


```python
svg_humdrum = vrv_render_page(1)
vrv_display_svg(svg_humdrum)
```

## 2. Convert MusicXML to MEI

We will use a Schubert Winterreise song from the Winterreise Dataset.


```python
musicxml_url = "https://raw.githubusercontent.com/piasteuck/winterreise-analysis/refs/heads/main/Schubert_Winterreise_Dataset_v2-0/01_RawData/score_musicxml/Schubert_D911-08.xml"

print("Converting MusicXML file...")
# Convert directly from URL
mei_output_xml = vrv_convert_to_mei(musicxml_url, is_url=True)

print(f"Conversion successful. Length of MEI: {len(mei_output_xml)} characters")
```

### Verify Render (MusicXML Source)


```python
svg_xml = vrv_render_page(1)
vrv_display_svg(svg_xml)
```

## 3. Save to File

You can save the output strings to `.mei` files.


```python
output_dir = "converted_mei"
os.makedirs(output_dir, exist_ok=True)

path_hum = os.path.join(output_dir, "beethoven_sonata_29_1.mei")
with open(path_hum, "w", encoding="utf-8") as f:
    f.write(mei_output_humdrum)

path_xml = os.path.join(output_dir, "schubert_winterreise_08.mei")
with open(path_xml, "w", encoding="utf-8") as f:
    f.write(mei_output_xml)
    
print(f"Files saved to:\n- {path_hum}\n- {path_xml}")
```
