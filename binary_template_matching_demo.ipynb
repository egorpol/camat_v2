
# Binary Template Matching â€” Demo Notebook (Methods 1â€“6) + Strided Convolution

This notebook lets you:
- Generate **synthetic** binary matrices `A` (big) and `B` (template), or **load your own** from files.
- Run and visualize six matching methods:
  1. Sliding **Hamming similarity**
  2. **Normalized cross-correlation**
  3. **Masked** Hamming (don't-care cells)
  4. **Exact** hit-or-miss (binary equality)
  5. **Distance-transform** (chamfer-like) score
  6. **Bit-packed Hamming** (POPCNT demo)
- Compare with a **classic 2D convolution/correlation** that supports a configurable **stride**.

> Heatmaps: higher is better, except Method 4 which returns a 0/1 exact-match map.



## What is the W mask doing?

`W` is a **binary mask** for the template `B` used in *masked Hamming* (Method 3).
- Where `W[y, x] = 1`, the cell is **compared**.
- Where `W[y, x] = 0`, the cell is **ignored** (a "don't care").

The masked Hamming distance over a placement is:
\[ D = |W| + |A \cdot W| - 2\,|A \wedge B \wedge W|, \]
and the similarity we plot is \( S = 1 - D/|W| \). This is helpful when parts of `B` are unreliable, occluded, or
you only care about specific positions in the template.



## Why is there a rainbow square on A?

That was just the rectangle outline being drawn with **four separate `plot` calls** that cycled through Matplotlib's default color cycle (blue, orange, green, red). In this version, the planted box is drawn as **one red rectangle** so it doesn't look rainbowy.


```python
# =====================
# CONFIG — edit me
# =====================
CFG = dict(
    # Choose: "synthetic" or "file"
    input_mode="file",
# - Where `W[y, x] = 0`, the cell is **ignored** (a "don't care").

# The masked Hamming distance over a placement is:
# \[ D = |W| + |A \cdot W| - 2\,|A \wedge B \wedge W|, \]
# and the similarity we plot is \( S = 1 - D/|W| \). This is helpful when parts of `B` are unreliable, occluded, or
# you only care about specific positions in the template.

    plant_template=True,   # embed B into A at a known location
    plant_yx=(None, None), # (y,x) or (None,None) for random
    plant_noise_flip_prob=0.0,
    mask_density=0.80,     # for Method 3 mask (1.0 = no don't-cares)
    seed=7,

    # --- File input options ---
    path_A="CAMAT_revamped/exports/binary_matrix_df_view_20250921_022343.npy",             # e.g. "A.npy" or "A.csv" or "A.png"
    path_B="CAMAT_revamped/exports/binary_matrix_prototype_view_20250921_022343.npy",             # e.g. "B.npy" or "B.csv" or "B.png"
    threshold_A=0.5,       # used if inputs are not already binary
    threshold_B=0.5,
    invert_A=False,        # set True if "foreground" is 0
    invert_B=False,

    # --- Strided convolution reference ---
    stride=(1, 1),         # (sy, sx)
    use_convolution=True,  # True = convolution (flip B); False = correlation

    # --- Visualization ---
    plot_backend="bokeh",  # "matplotlib" or "bokeh"
)
print("Loaded CFG:", CFG)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

try:
    from scipy.signal import correlate2d
    from scipy import ndimage as ndi
    SCIPY_OK = True
except Exception as e:
    SCIPY_OK = False
    print("SciPy not available; some methods will fallback/skip. Error:", e)


class Plotter:
    """Lightweight plotting facade that supports Matplotlib and Bokeh backends."""

    def __init__(self, backend: str = "matplotlib"):
        backend = (backend or "matplotlib").lower()
        self.backend = backend
        if backend == "matplotlib":
            import matplotlib.pyplot as _plt
            from matplotlib import patches as _patches

            self._plt = _plt
            self._patches = _patches
        elif backend == "bokeh":
            try:
                from bokeh.io import output_notebook, show
                from bokeh.plotting import figure
                from bokeh.models import LinearColorMapper, ColorBar, BasicTicker
                from bokeh.palettes import Greys256, Viridis256
            except Exception as err:
                raise RuntimeError(f"Bokeh backend requested but it is unavailable: {err}")

            output_notebook(hide_banner=True)
            self._show = show
            self._figure = figure
            self._LinearColorMapper = LinearColorMapper
            self._ColorBar = ColorBar
            self._BasicTicker = BasicTicker
            self._palette_binary = list(Greys256)[::-1]
            self._palette_heatmap = list(Viridis256)
        else:
            raise ValueError(f"Unknown plotting backend: {backend}")

    @staticmethod
    def _figure_dimensions(shape):
        h, w = shape
        width = min(max(int(w * 6), 320), 900)
        height = min(max(int(h * 6), 320), 900)
        return width, height

    def show_binary(self, img, title, highlight=None):
        if self.backend == "matplotlib":
            fig, ax = self._plt.subplots()
            ax.imshow(img, interpolation="nearest")
            ax.set_title(title)
            ax.axis("off")
            if highlight is not None:
                y, x, h, w = highlight
                rect = self._patches.Rectangle((x, y), w-1, h-1, fill=False, edgecolor="red", linewidth=2)
                ax.add_patch(rect)
            return fig

        img = np.asarray(img, dtype=float)
        width, height = self._figure_dimensions(img.shape)
        mapper = self._LinearColorMapper(palette=self._palette_binary, low=0.0, high=1.0)
        fig = self._figure(
            title=title,
            x_range=(0, img.shape[1]),
            y_range=(img.shape[0], 0),
            width=width,
            height=height,
            toolbar_location="right",
            tools="pan,wheel_zoom,box_zoom,reset,save",
        )
        fig.image(image=[img], x=0, y=0, dw=img.shape[1], dh=img.shape[0], color_mapper=mapper)
        if highlight is not None:
            y, x, h, w = highlight
            xc = x + w / 2 - 0.5
            yc = y + h / 2 - 0.5
            fig.rect(xc, yc, width=w, height=h, fill_alpha=0.0, line_color="red", line_width=2)
        fig.axis.visible = False
        fig.grid.visible = False
        self._show(fig)
        return fig

    def show_heatmap(self, data, title, best=None):
        if self.backend == "matplotlib":
            fig, ax = self._plt.subplots()
            im = ax.imshow(data, interpolation="nearest")
            ax.set_title(title)
            ax.axis("off")
            if best is not None:
                y, x = best
                ax.plot(x, y, marker="x", markersize=10, markeredgewidth=2, color="red")
            fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
            return fig

        data = np.asarray(data, dtype=float)
        width, height = self._figure_dimensions(data.shape)
        low = float(np.nanmin(data))
        high = float(np.nanmax(data))
        if np.isclose(high, low):
            high = low + 1e-6
        mapper = self._LinearColorMapper(palette=self._palette_heatmap, low=low, high=high)
        fig = self._figure(
            title=title,
            x_range=(0, data.shape[1]),
            y_range=(data.shape[0], 0),
            width=width,
            height=height,
            toolbar_location="right",
            tools="pan,wheel_zoom,box_zoom,reset,save",
        )
        fig.image(image=[data], x=0, y=0, dw=data.shape[1], dh=data.shape[0], color_mapper=mapper)
        if best is not None:
            y, x = best
            fig.cross(x + 0.5, y + 0.5, size=18, line_color="red", line_width=2)
        color_bar = self._ColorBar(color_mapper=mapper, ticker=self._BasicTicker(desired_num_ticks=6))
        fig.add_layout(color_bar, "right")
        fig.grid.visible = False
        fig.axis.visible = False
        self._show(fig)
        return fig

    def flush(self):
        if self.backend == "matplotlib":
            self._plt.show()


# ---------- I/O helpers ----------
def _to_gray01(arr):
    arr = np.asarray(arr)
    if arr.ndim == 3:
        arr = arr.mean(axis=2)  # average RGB
    arr = arr.astype(np.float64)
    if arr.max() > 1.5:
        arr = arr / 255.0
    return np.clip(arr, 0.0, 1.0)


def load_matrix(path, threshold=0.5, invert=False):
    p = Path(path)
    if not p.exists():
        raise FileNotFoundError(f"No such file: {path}")
    ext = p.suffix.lower()
    if ext == ".npy":
        X = np.load(p)
    elif ext == ".npz":
        z = np.load(p)
        key = list(z.keys())[0]
        X = z[key]
    elif ext in [".csv", ".txt"]:
        X = np.genfromtxt(p, dtype=float)
    elif ext in [".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff"]:
        X = plt.imread(p)
        X = _to_gray01(X)
    else:
        raise ValueError(f"Unsupported extension: {ext}")
    X = np.asarray(X)
    if X.dtype != np.bool_ and not np.array_equal(X, X.astype(bool)):
        Xb = (X >= threshold).astype(np.uint8)
    else:
        Xb = X.astype(np.uint8)
    if invert:
        Xb = 1 - Xb
    return Xb


# ---------- Integral image ----------
def integral_image(x: np.ndarray) -> np.ndarray:
    ii = np.cumsum(np.cumsum(x.astype(np.int64), axis=0), axis=1)
    ii = np.pad(ii, ((1, 0), (1, 0)), mode="constant")
    return ii


def box_sum(ii: np.ndarray, hh: int, ww: int) -> np.ndarray:
    return ii[hh:, ww:] - ii[:-hh, ww:] - ii[hh:, :-ww] + ii[:-hh, :-ww]
```

```python

# Build A, B, and optional mask W. If using file mode, mask will be dense (all ones).
cfg = CFG
rng = np.random.default_rng(cfg['seed'])

if cfg['input_mode'] == 'file':
    assert cfg['path_A'] and cfg['path_B'], "Provide path_A and path_B for file mode."
    A = load_matrix(cfg['path_A'], threshold=cfg['threshold_A'], invert=cfg['invert_A'])
    B = load_matrix(cfg['path_B'], threshold=cfg['threshold_B'], invert=cfg['invert_B'])
    h, w = B.shape
    H, W = A.shape
    plant_loc = None
    Wmask = np.ones_like(B, dtype=np.uint8)
else:
    H, W = cfg['H'], cfg['W']
    h, w = cfg['h'], cfg['w']
    if h > H or w > W:
        raise ValueError("Template B must be smaller than A (h<=H and w<=W).")
    A = (rng.random((H, W)) < cfg['density_A']).astype(np.uint8)
    B = (rng.random((h, w)) < cfg['density_B']).astype(np.uint8)

    plant_loc = None
    if cfg['plant_template']:
        y0, x0 = cfg['plant_yx']
        if y0 is None:
            y0 = int(rng.integers(0, H - h + 1))
        if x0 is None:
            x0 = int(rng.integers(0, W - w + 1))
        plant_loc = (y0, x0)
        B_planted = B.copy()
        if cfg['plant_noise_flip_prob'] > 0:
            noise = rng.random((h, w)) < cfg['plant_noise_flip_prob']
            B_planted = np.bitwise_xor(B_planted, noise.astype(np.uint8))
        A[y0:y0+h, x0:x0+w] = B_planted

    if cfg['mask_density'] >= 1.0:
        Wmask = np.ones((h, w), dtype=np.uint8)
    else:
        Wmask = (rng.random((h, w)) < cfg['mask_density']).astype(np.uint8)

print(f"A shape: {A.shape}, B shape: {B.shape}, mask ones: {int(Wmask.sum())}/{Wmask.size}")

try:
    plotter = Plotter(cfg.get('plot_backend', 'matplotlib'))
except RuntimeError as err:
    print(err)
    print("Falling back to Matplotlib backend.")
    plotter = Plotter("matplotlib")
```

```python

# -------- Methods implementation (same as before, omitted comments for brevity) --------

def match_hamming(A, B):
    H, W = A.shape
    h, w = B.shape
    if SCIPY_OK:
        and_count = correlate2d(A, B[::-1, ::-1], mode='valid')
    else:
        out = np.zeros((H - h + 1, W - w + 1), dtype=np.int64)
        for y in range(H - h + 1):
            block = A[y:y+h]
            for x in range(W - w + 1):
                out[y, x] = np.sum(block[:, x:x+w] * B)
        and_count = out
    Aii = integral_image(A)
    Awin = box_sum(Aii, h, w)
    Bsum = int(B.sum())
    D = Bsum + Awin - 2 * and_count
    return 1.0 - (D / max(Bsum, 1))

def match_normalized_corr(A, B):
    if not SCIPY_OK:
        raise RuntimeError("SciPy required for Method 2 (correlate2d).")
    h, w = B.shape
    Bsum = float(B.sum())
    and_count = correlate2d(A, B[::-1, ::-1], mode='valid').astype(np.float64)
    Aii = integral_image(A)
    Awin = box_sum(Aii, h, w).astype(np.float64)
    denom = np.sqrt(np.maximum(Awin * max(Bsum, 1.0), 1e-12))
    return np.clip(and_count / denom, 0.0, 1.0)

def match_masked_hamming(A, B, Wmask):
    if not SCIPY_OK:
        raise RuntimeError("SciPy required for Method 3 (masked correlate2d).")
    BW = (B & Wmask).astype(np.float64)
    Wflip = Wmask[::-1, ::-1].astype(np.float64)
    BWflip = BW[::-1, ::-1]
    and_count = correlate2d(A, BWflip, mode='valid')
    A_in_mask = correlate2d(A, Wflip, mode='valid')
    Wsum = float(Wmask.sum())
    D = Wsum + A_in_mask - 2.0 * and_count
    return 1.0 - (D / max(Wsum, 1.0))

def match_exact(A, B):
    if SCIPY_OK:
        and_count = correlate2d(A, B[::-1, ::-1], mode='valid')
        Aii = integral_image(A)
        Awin = box_sum(Aii, *B.shape)
        Bsum = int(B.sum())
        D = Bsum + Awin - 2 * and_count
        return (D == 0).astype(float)
    else:
        H, W = A.shape
        h, w = B.shape
        S = np.zeros((H - h + 1, W - w + 1), dtype=float)
        for y in range(H - h + 1):
            for x in range(W - w + 1):
                S[y, x] = float(np.array_equal(A[y:y+h, x:x+w], B))
        return S

def match_distance_transform(A, B):
    if not SCIPY_OK:
        raise RuntimeError("SciPy required for Method 5 (distance transform).")
    dist_to_ones = ndi.distance_transform_edt(A == 0).astype(np.float64)
    B1 = (B == 1).astype(np.float64)
    sumdist = correlate2d(dist_to_ones, B1[::-1, ::-1], mode='valid')
    mn, mx = float(sumdist.min()), float(sumdist.max())
    return 1.0 - (sumdist - mn) / (mx - mn + 1e-12)

POPCOUNT_TABLE = np.array([bin(i).count('1') for i in range(256)], dtype=np.uint8)

def popcount_u64(arr_u64):
    b = arr_u64.view(np.uint8)
    return int(POPCOUNT_TABLE[b].sum())

def pack_rows_u64(X):
    H, W = X.shape
    pad_bits = (-W) % 64
    Xpad = np.pad(X.astype(np.uint8), ((0, 0), (0, pad_bits)), mode='constant')
    packed_bytes = np.packbits(Xpad, axis=1, bitorder='little')
    nbytes = packed_bytes.shape[1]
    pad_bytes = (-nbytes) % 8
    if pad_bytes:
        packed_bytes = np.pad(packed_bytes, ((0, 0), (0, pad_bytes)), mode='constant')
    X_u64 = packed_bytes.view(np.uint64)
    return X_u64, W + pad_bits

def _hamming_bitpacked_row(A_u64_row, x, B_u64_row):
    wwords = B_u64_row.shape[0]
    word_off = x // 64
    bit_off = x % 64
    src = A_u64_row[word_off: word_off + wwords + 1]
    if src.shape[0] < wwords + 1:
        src = np.pad(src, (0, wwords + 1 - src.shape[0]))
    if bit_off == 0:
        win = src[:wwords]
    else:
        low = src[:wwords]
        high = src[1:wwords+1]
        win = (low >> bit_off) | (high << (64 - bit_off))
    xor = np.bitwise_xor(win, B_u64_row)
    return popcount_u64(xor)

def match_bitpacked_hamming(A, B):
    H, W = A.shape
    h, w = B.shape
    A_u64, _ = pack_rows_u64(A)
    B_u64, _ = pack_rows_u64(B)
    out = np.empty((H - h + 1, W - w + 1), dtype=float)
    total_bits = h * w
    for y in range(H - h + 1):
        rows = A_u64[y:y+h]
        for x in range(W - w + 1):
            d = 0
            for i in range(h):
                d += _hamming_bitpacked_row(rows[i], x, B_u64[i])
            out[y, x] = 1.0 - d / max(total_bits, 1)
    return out

def conv2d_strided(A, B, stride=(1, 1), use_convolution=True):
    sy, sx = stride
    K = B[::-1, ::-1] if use_convolution else B
    if SCIPY_OK:
        full = correlate2d(A, K, mode='valid')
    else:
        H, W = A.shape
        h, w = B.shape
        out = np.zeros((H - h + 1, W - w + 1), dtype=float)
        for y in range(H - h + 1):
            block = A[y:y+h]
            for x in range(W - w + 1):
                out[y, x] = float((block[:, x:x+w] * K).sum())
        full = out
    return full[::sy, ::sx].astype(float)
```

```python
# Visualize inputs
highlight = None if plant_loc is None else (plant_loc[0], plant_loc[1], B.shape[0], B.shape[1])
plotter.show_binary(A, "A (big binary matrix)", highlight=highlight)
plotter.show_binary(B, "B (template)")
plotter.show_binary(Wmask, "W (mask for Method 3)")

# Method 1
S1 = match_hamming(A, B)
y1, x1 = np.unravel_index(np.argmax(S1), S1.shape)
plotter.show_heatmap(S1, f"Method 1: Hamming similarity (best @ ({y1},{x1}) = {S1[y1,x1]:.3f})", best=(y1, x1))

# Method 2
if SCIPY_OK:
    S2 = match_normalized_corr(A, B)
    y2, x2 = np.unravel_index(np.argmax(S2), S2.shape)
    plotter.show_heatmap(S2, f"Method 2: Normalized cross-correlation (best @ ({y2},{x2}) = {S2[y2,x2]:.3f})", best=(y2, x2))
else:
    print("Skipping Method 2 (SciPy not available).")

# Method 3
if SCIPY_OK:
    S3 = match_masked_hamming(A, B, Wmask)
    y3, x3 = np.unravel_index(np.argmax(S3), S3.shape)
    plotter.show_heatmap(S3, f"Method 3: Masked Hamming (best @ ({y3},{x3}) = {S3[y3,x3]:.3f})", best=(y3, x3))
else:
    print("Skipping Method 3 (SciPy not available).")

# Method 4
S4 = match_exact(A, B)
hits = int(S4.sum())
y4, x4 = (np.unravel_index(np.argmax(S4), S4.shape) if hits > 0 else (None, None))
plotter.show_heatmap(S4, f"Method 4: Exact hit-or-miss (hits={hits})", best=(y4, x4) if hits > 0 else None)

# Method 5
if SCIPY_OK:
    S5 = match_distance_transform(A, B)
    y5, x5 = np.unravel_index(np.argmax(S5), S5.shape)
    plotter.show_heatmap(S5, f"Method 5: Distance-transform (best @ ({y5},{x5}) = {S5[y5,x5]:.3f})", best=(y5, x5))
else:
    print("Skipping Method 5 (SciPy not available).")

# Method 6
S6 = match_bitpacked_hamming(A, B)
y6, x6 = np.unravel_index(np.argmax(S6), S6.shape)
plotter.show_heatmap(S6, f"Method 6: Bit-packed Hamming (best @ ({y6},{x6}) = {S6[y6,x6]:.3f})", best=(y6, x6))

print("=== Best locations summary ===")
print(f"Method 1 (Hamming):        ({y1},{x1}) score={S1[y1,x1]:.4f}")
if SCIPY_OK:
    print(f"Method 2 (Norm. corr):     ({y2},{x2}) score={S2[y2,x2]:.4f}")
    print(f"Method 3 (Masked Hamming): ({y3},{x3}) score={S3[y3,x3]:.4f}")
else:
    print("Method 2: skipped; Method 3: skipped (no SciPy)")
print(f"Method 4 (Exact):          hits={hits}" + (f", first @ ({y4},{x4})" if hits > 0 else ""))
if SCIPY_OK:
    print(f"Method 5 (DT score):       ({y5},{x5}) score={S5[y5,x5]:.4f}")
else:
    print("Method 5: skipped (no SciPy)")
print(f"Method 6 (Bit-pack Ham):   ({y6},{x6}) score={S6[y6,x6]:.4f}")
plotter.flush()
```

```python
# ---- Classic convolution / correlation with stride ----
Sconv = conv2d_strided(A, B, stride=CFG['stride'], use_convolution=CFG['use_convolution'])
yy, xx = np.unravel_index(np.argmax(Sconv), Sconv.shape)
mode_name = "Convolution" if CFG['use_convolution'] else "Cross-correlation"
plotter.show_heatmap(Sconv, f"{mode_name} with stride={CFG['stride']} (best @ ({yy},{xx}) = {Sconv[yy,xx]:.3f})", best=(yy, xx))
plotter.flush()
```
