# Parser Comparison: Bach, BWV 80

Contrast the music21 and partitura parsing backends on the MEI sample [`Bach-JS_Ein_feste_Burg.mei`](https://raw.githubusercontent.com/music-encoding/sample-encodings/main/MEI_5.0/Music/Complete_examples/Bach-JS_Ein_feste_Burg.mei). Both runs share identical configuration so that differences highlight parser behavior only.

```python
import pandas as pd
from IPython.display import display

from py_scripts.parser_registry import get_parse_files
```

```python
FILE_SOURCES = [
    'https://raw.githubusercontent.com/music-encoding/sample-encodings/main/MEI_5.0/Music/Complete_examples/Bach-JS_Ein_feste_Burg.mei'
]

FILTER_ZERO_DURATION = True
ADJUST_FRACTIONAL_DURATION = True
PARSE_ENHARMONIC = True
PLOTTING_BACKEND = 'bokeh'
SHOW_MEASURE_LINES = True
MEASURE_LINE_COLOR = 'black'
HOVER_ENABLED = True
HOVER_FIELDS = ['measure', 'local_onset', 'global_onset', 'duration', 'pitch', 'pitch_enharmonic', 'midi', 'voice', 'xml_id']
DISPLAY_PREVIEW = True
PREVIEW_ROWS = 20
CLEANUP_REMOTE = True
RETURN_PLOTS = False
STRIP_TIES = True
ALIGN_ACCIDENT_SCHEMA = True
COLORIZE_VOICES = True
PALETTE = 'Category20'
INCLUDE_XML_IDS = True
```

```python
def run_backend(name: str):
    parse_files = get_parse_files(name)
    print(f'=== Parsing with {name} backend ===')
    results, dfs_by_name, df_processed = parse_files(
        FILE_SOURCES,
        filter_zero_duration=FILTER_ZERO_DURATION,
        adjust_fractional_duration=ADJUST_FRACTIONAL_DURATION,
        parse_enharmonic=PARSE_ENHARMONIC,
        backend=PLOTTING_BACKEND,
        show_measure_lines=SHOW_MEASURE_LINES,
        measure_line_color=MEASURE_LINE_COLOR,
        show_hover=HOVER_ENABLED,
        hover_fields=HOVER_FIELDS,
        display_preview=DISPLAY_PREVIEW,
        preview_rows=PREVIEW_ROWS,
        cleanup_remote=CLEANUP_REMOTE,
        return_plots=RETURN_PLOTS,
        plot_width=900,
        plot_height=600,
        zoom_drag_dim='both',
        zoom_wheel_dim='width',
        strip_ties=STRIP_TIES,
        align_accident_schema=ALIGN_ACCIDENT_SCHEMA,
        colorize_voices=COLORIZE_VOICES,
        palette=PALETTE,
        include_xml_ids=INCLUDE_XML_IDS,
    )
    return results, dfs_by_name, df_processed

backends = ['music21', 'partitura']
run_results = {}
for backend in backends:
    run_results[backend] = run_backend(backend)
```

```python
def summarize_df(df: pd.DataFrame) -> pd.Series:
    cols = ['Measure', 'Local Onset', 'Global Onset', 'Duration', 'Pitch', 'MIDI', 'Voice']
    available = [c for c in cols if c in df.columns]
    summary = {
        'rows': len(df),
        'unique_midi': df['MIDI'].nunique() if 'MIDI' in df.columns else None,
        'min_duration': float(df.loc[df['Duration'] > 0, 'Duration'].min()) if 'Duration' in df.columns else None,
        'voices': df['Voice'].nunique() if 'Voice' in df.columns else None,
        'has_xml_id': bool(df.get('xml_id') is not None and df['xml_id'].notna().any())
    }
    return pd.Series(summary)

summary_rows = {}
for backend, (_, _, df_proc) in run_results.items():
    summary_rows[backend] = summarize_df(df_proc)

comparison_df = pd.DataFrame(summary_rows)
display(comparison_df)
```

```python
def compare_dataframes(df_a: pd.DataFrame, df_b: pd.DataFrame, *, label_a: str, label_b: str):
    common_cols = sorted(set(df_a.columns) & set(df_b.columns))
    df_a_common = df_a[common_cols].reset_index(drop=True)
    df_b_common = df_b[common_cols].reset_index(drop=True)

    shape_info = {
        label_a: df_a_common.shape,
        label_b: df_b_common.shape,
    }
    print('Shapes:', shape_info)

    equal = df_a_common.equals(df_b_common)
    print(f'Exact match on common columns: {equal}')

    if not equal:
        merged = df_a_common.merge(
            df_b_common,
            how='outer',
            indicator=True,
        )
        print('Rows unique to', label_a, len(merged[merged['_merge'] == 'left_only']))
        print('Rows unique to', label_b, len(merged[merged['_merge'] == 'right_only']))

        try:
            diff = df_a_common.compare(df_b_common)
            display(diff.head())
        except ValueError:
            print('Unable to align rows for DataFrame.compare; inspect merged output instead.')

music21_df = run_results['music21'][2]
partitura_df = run_results['partitura'][2]
compare_dataframes(music21_df, partitura_df, label_a='music21', label_b='partitura')
```

## Suggested comparison tactics
- `DataFrame.compare(other)` highlights cell-level differences after aligning indices; best used when both parsers emit the same row count/order.
- `DataFrame.equals(other)` or boolean masks help quickly confirm identical subsets (for example, equality across `['Global Onset', 'Duration', 'MIDI']`).
- Set-style operations such as `df_a.merge(df_b, how='outer', indicator=True)` reveal notes unique to one backend after selecting a robust key.
- Aggregate statistics (row counts, duration distributions, number of voices) offer a higher-level contrast when note-level alignment diverges.
- When enharmonic spelling differs, normalizing columns with `ALIGN_ACCIDENT_SCHEMA` or comparing MIDI numbers avoids false positives.
